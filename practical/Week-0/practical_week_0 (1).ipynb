Program 2: Operators
Write a Python program to demonstrate arithmetic operators.
Write a program to demonstrate comparison operators using two numbers.
Write a program to demonstrate logical operators (and, or, not).
Write a program to determine whether a student is eligible for placement based on CGPA and attendance using logical operators.


```python
a = 10
b = 3

print(f"a + b = {a + b}")  # Addition
print(f"a - b = {a - b}")  # Subtraction
print(f"a * b = {a * b}")  # Multiplication
print(f"a / b = {a / b}")  # Division
print(f"a % b = {a % b}")  # Modulus
print(f"a ** b = {a ** b}") # Exponentiation
print(f"a // b = {a // b}") # Floor Division
```

```python
p = True
q = False

print(f"p and q: {p and q}") # Logical AND
print(f"p or q: {p or q}")   # Logical OR
print(f"not p: {not p}")     # Logical NOT
```

```python
# Define eligibility criteria
MIN_CGPA = 3.0
MIN_ATTENDANCE_PERCENTAGE = 75

# Student details
student_cgpa = 3.2
student_attendance = 80 # in percentage

# Check eligibility using logical operators
is_eligible = (student_cgpa >= MIN_CGPA) and (student_attendance >= MIN_ATTENDANCE_PERCENTAGE)

print(f"Student CGPA: {student_cgpa}")
print(f"Student Attendance: {student_attendance}%")

if is_eligible:
    print("Congratulations! The student is eligible for placement.")
else:
    print("The student is not eligible for placement.")

# Another example
student_cgpa_2 = 2.8
student_attendance_2 = 90

is_eligible_2 = (student_cgpa_2 >= MIN_CGPA) and (student_attendance_2 >= MIN_ATTENDANCE_PERCENTAGE)

print(f"\nStudent CGPA: {student_cgpa_2}")
print(f"Student Attendance: {student_attendance_2}%")

if is_eligible_2:
    print("Congratulations! The student is eligible for placement.")
else:
    print("The student is not eligible for placement.")
```

Program 4: Lists and Indexing
Write a Python program to create a list and access its elements using indexing.
Write a program to display the first, second, and last elements of a list.
Write a program to demonstrate list slicing.
Write a program to add a new element to an existing list using the append() method.
Write a program to display the total number of elements in a list.

```python
# 1.
fruits = ["apple", "banana", "cherry", "date", "elderberry"]
print("Original List:", fruits)

# 2.
first_element = fruits[0]
second_element = fruits[1]
last_element = fruits[-1]

print("\nFirst Element:", first_element)
print("Second Element:", second_element)
print("Last Element:", last_element)

# 3.
print("\nList Slicing")
print("First three elements (0:3):", fruits[0:3])
print("Elements from index 2 to end (2:):", fruits[2:])
print("Middle elements (1:4):", fruits[1:4])

# 4.
fruits.append("Mango")
print("\nAfter Adding an Element ")
print("Updated List:", fruits)

# 5.
total_elements = len(fruits)
print("\nList Length")
print("Total number of elements:", total_elements)
```

Program 5: Dictionaries
Write a Python program to create a dictionary and display all its key-value pairs.
Write a program to access the value associated with a specific key in a dictionary.
Write a program to iterate through a dictionary using the items() method.
Write a program to store different machine learning models in a dictionary and display their names and objects.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

# 1.
student = {
    "name": "Aryan",
    "age": 19,
    "course": "Machine Learning",
    "gpa": 9.9
}
print("Complete Dictionary:", student)

# 2.
student_name = student["name"]
student_gpa = student.get("gpa")

print("\nAccessing Specific Keys")
print("Student Name:", student_name)
print("Student GPA:", student_gpa)

# 3.
print("\nIterating using items()")
for key, value in student.items():
    print(f"Key: {key:<10} | Value: {value}")

# 4.
ml_models = {
    "Logistic Regression": LogisticRegression(),
    "Decision Tree": DecisionTreeClassifier(),
    "Random Forest": RandomForestClassifier()
}

print("\nMachine Learning Models Dictionary")
for model_name, model_obj in ml_models.items():
    print(f"Model Name: {model_name:<20} | Object: {model_obj}")
```

Program 6: Conditional Statements
Write a Python program to classify a student's CGPA into High, Mid, or Low categories using if-elif-else.
Write a program to check whether missing values are present in a dataset using conditional statements.
Write a program to determine whether a student is eligible for placement based on CGPA.
Write a program that uses if-elif-else to assign grades based on marks.


```python
# 1
cgpa = 8.2

if cgpa >= 8.0:
    print("CGPA Category: High")
elif cgpa >= 6.5:
    print("CGPA Category: Mid")
else:
    print("CGPA Category: Low")


# 2
missing_count = 5

if missing_count > 0:
    print(f"Missing values found: {missing_count}")
else:
    print("No missing values found.")


# 3
student_cgpa = 7.5
min_cgpa_required = 7.0

if student_cgpa >= min_cgpa_required:
    print("Eligible for placement")
else:
    print("Not eligible for placement")


# 4
marks = 85

if marks >= 90:
    print("Grade: A+")
elif marks >= 80:
    print("Grade: A")
elif marks >= 70:
    print("Grade: B")
elif marks >= 60:
    print("Grade: C")
elif marks >= 50:
    print("Grade: D")
else:
    print("Grade: F")
```

Program 7: Loops
Write a Python program to display all elements in a list using a for loop.
Write a program to calculate the average of numbers stored in a list.
Write a program to display only the values greater than a given threshold using a loop.
Write a program to find the maximum and minimum values in a list using loops.
Write a program to count the number of elements satisfying a given condition using a loop.

```python
# 1
numbers = [12, 45, 67, 23, 89, 34]
for num in numbers:
    print(num)


# 2
numbers = [10, 20, 30, 40, 50]
total = 0
for num in numbers:
    total += num
average = total / len(numbers)
print("Average:", average)


# 3
numbers = [15, 82, 25, 47, 90, 61, 33]
threshold = 40
print("Values greater than threshold:")
for num in numbers:
    if num > threshold:
        print(num)


# 4
numbers = [42, 17, 88, 5, 63, 91, 29]
max_val = numbers[0]
min_val = numbers[0]

for num in numbers:
    if num > max_val:
        max_val = num
    if num < min_val:
        min_val = num

print("Maximum value:", max_val)
print("Minimum value:", min_val)


# 5
numbers = [12, 7, 24, 15, 30, 9, 18, 41]
count = 0
for num in numbers:
    if num % 2 == 0:
        count += 1

print("Count of even numbers:", count)
```
